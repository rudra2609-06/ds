### Find the magnitude of the vector C = [2, -3, 1].

```python
v = [2,-3,1]

np.linalg.norm(v)
```

### Given the vectors D = [1, 2, 3] and E = [4, 5, 6], calculate the cosine of the angle between them.

```python
d = [1,2,3]
e = [4,5,6]

angle_of_cosine = np.dot(d,e) / (np.linalg.norm(d) * np.linalg.norm(e))
angle_of_cosine
```

### Given the vectors F = [1, 2, 3] and G = [4, 5, 6], calculate the cross product of the two vectors.

```python
f = [1,2,3]
g = [4,5,6]

np.cross(f,g)
```
