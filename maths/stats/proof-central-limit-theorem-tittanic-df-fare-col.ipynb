## Performing Central Limit Theorem On Fare Column


```python
fare = df['Fare']
```

```python
import seaborn as sns
```

```python
sns.displot(data = fare, kind = 'kde')
```

```python
# if i just do fare.mean() it will give me population mean directly 
# but we will remove the population mean from sample using "CLT"
```

```python
samples = []

for i in range(100):
    samples.append(df['Fare'].sample(50).values.tolist())
    
```

```python
samples
```

```python
means = np.mean(samples,axis=1)
```

```python
sns.displot(data = means, kind = 'kde')
```

```python
np.mean(means)
```
