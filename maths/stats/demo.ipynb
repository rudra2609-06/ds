```python
percentiles = np.percentile(df['fare'].values,[25,50,75])
```

```python
iqr = percentiles[2] - percentiles[0]
minimum = percentiles[0] - 1.5 * iqr
maximum = percentiles[2] + 1.5 * iqr
minimum
```

```python
outliers = df[(df['fare'] > maximum) | (df['fare'] < minimum)]

```
