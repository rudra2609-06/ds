```python
import numpy as np
import pandas as pd
import scipy.stats as stats
```

# Example 1
***A researcher wants to investigate if there is an association b/w the level of education (categorical variable) and the preference of particular exercise (categorical variable) among the group of 150 individuals. The researcher collects the following sample and creates contingency table***

| Education        | Yoga | Running | Swimming | Total |
|------------------|-----:|--------:|---------:|------:|
| High School      | 15   | 20      | 10       | 45    |
| Bachelor's       | 20   | 30      | 15       | 65    |
| Master's or PhD  | 5    | 15      | 20       | 40    |
| **Total**        | **40** | **65** | **45** | **150** |

## Test Starts

```python
# H0 -> There is significant association b/w Education level and preference for exercise
# H1 -> There is no significant association b/w Education level and preference for exercise
```

```python
observed = pd.DataFrame(
    {
        "Yoga": [15, 20, 5],
        "Running": [20, 30, 15],
        "Swimming": [10, 15, 20],
        "Total": [45, 65, 40]
    },
    index=["High School", "Bachelor's", "Master's or PhD"]
)

observed.loc["Total"] = observed.sum()

observed
```

```python


# Row totals
row_totals = observed.sum(axis=1)

# Column totals
col_totals = observed.sum(axis=0)

# Grand total
grand_total = observed.sum().sum()

# Expected table
expected = pd.DataFrame(
    [[row_totals[i] * col_totals[j] / grand_total
      for j in range(len(col_totals))]
     for i in range(len(row_totals))],
    index=observed.index,
    columns=observed.columns
)

expected
```

```python
# you need not to calculate the expected table it does on its own
res = stats.chi2_contingency(observed)
res
```

```python
if res.pvalue > 0.05:
	print('Null Hypothesis Cannot Be Rejected')
else:
	print('Null Hypothesis Can Be Rejected')
```
