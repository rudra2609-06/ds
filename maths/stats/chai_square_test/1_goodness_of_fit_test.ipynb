# Example 1
***Suppose we have a six-sided fair die, and we want to test if the die is indeed fair.We roll the die 60 times and record the number of times each side comes up.We will use the chi-square Goodness of fit test to determine if the observed frequencies are consistent with a fair dice (i.e Uniform Dist)***

### Observed Frequencies:
- 1 -> 2 times
- 2 -> 14 times
- 3 -> 12 times
- 4 -> 10 times
- 5 -> 10 times
- 6 -> 12 times


```python
rolls = np.random.randint(1,7,size=60)

faces,counts = np.unique(rolls,return_counts=True)
```

```python
for i in range(len(faces)):
	print(f'{i + 1} -> {counts[i]}')
```

## Test Starts

```python
#  H0 -> dice is fair
#  h1 -> dice is not fair
```

```python
# if dice is fair which means it follows uniform dist therefore 
# expected for each -> 10 (1/6)
```

```python
observed = [2,14,12,10,10,12]
expected = [10] * 6

chi_square_statistic,p_value = stats.chisquare(f_obs=observed,f_exp=expected)
```

```python
if p_value > 0.05:
	print('Null Hypothesis Cannot be Rejected')
else:
	print('Null Hypothesis Rejected')

# Therefore dice is indeed fair
```
