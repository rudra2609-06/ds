```python
import pandas as pd
import numpy as np
import scipy.stats as stats
```

```python
df = pd.read_csv('Titanic-Dataset.csv').dropna()
```

```python
sample_male = df[df['Sex'] == 'male']['Age'].sample(25)
sample_females = df[df['Sex'] == 'female']['Age'].sample(25)
```

```python
# H0 -> Mean age of male and female is same
# H1 -> Mean age of male is higher than female
```

### Test Starts

```python
shapiro_male = stats.shapiro(sample_male) # -> 0.9 therefore follow ND
shapiro_female = stats.shapiro(sample_females) # -> 0.86 therefore follows ND
```

```python
levene_test = stats.levene(sample_male,sample_females)
levene_test # -> p_value also greater than 0.05 therefore variance same assumption assured
```

```python
t_stats,pvalue = stats.ttest_ind(sample_male,sample_females)
```

```python
if pvalue > 0.05:
	print("Rejected Null Hypothesis")
else:
	print("Can't Reject Null Hypothesis")
```
