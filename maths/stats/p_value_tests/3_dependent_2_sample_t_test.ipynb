```python
import pandas as pd
import numpy as np
import scipy.stats as stats
```

***Let's assume there is 8 week long significant weight loss program. And we have 15 participants enrolled in the program and measure their weights before and after the program. The goal is to test wheather the new weight loss program leads to a significant reduction in the participants weight***

***Before The Program:- [80,92,75,68,85,78,73,90,70,88,76,80,82,77,91]***

***After The Program:- [78,93,81,67,88,76,74,91,69,88,77,81,80,79,88]***

***Significance Level: 0.05***

```python
# H0 -> avg weight before = avg weight after
# H1 -> avg weight before > avg weight after -> weight loss occured
```

```python
before = np.array([80,92,75,68,85,78,73,90,70,88,76,80,82,77,91])
after = np.array([78,93,81,67,88,76,74,91,69,88,77,81,80,79,88])

diff = before - after
```

```python
res = stats.shapiro(diff)
res # -> this follow Normal distribution as p_value is 0.2
```

```python
t_statistic,p_value = stats.ttest_rel(before,after)
```

```python
p_value
```

```python
if p_value < 0.05:
	print('Rejected Null Hypothesis')
else:
	print("Can't Reject Null Hypothesis")
```
