# Task 1 Soln

```python
m1 = np.array([75, 80, 90, 85, 70, 75, 80, 85, 90, 95, 80, 75, 85, 90, 95, 80, 75, 70, 85, 90, 95, 75, 70, 80, 85])

m2 = np.array([80, 75, 85, 90, 95, 70, 75, 80, 85, 90, 85, 75, 80, 90, 95, 85, 75, 70, 80, 90, 85, 70, 75, 80, 85])

alpha = 0.05
```

```python
# H0 -> mean maths scores are equal
# H1 -> mean maths scores differ
```

#### Sub-Q1: Perform normality test and equal variance test for both schools.

```python
# normality test

res1 = stats.shapiro(m1) # its p-value is lesser than significance level
res2 = stats.shapiro(m2) # its p-value is greater than significance level
res1
```

```python
# equal variance test

res_var_1 = stats.levene(m1,m2)
res_var_1 # -> p-value is greater than alpha therefore their variance are equal
```

## Ans of Sub-Q1:

Normality is violated for the first school, while the second school's data generally resembles a normal distribution. Both schools pass Levene's test, indicating equal variances.

#### Sub-Q2: Is there a significant difference between the mean math scores of the two schools?

```python
t_statistic,p_value = stats.ttest_ind(m1,m2)
p_value
```

```python
if p_value > alpha:
	print("Null Hypothesis Can't be rejected")
else:
	print("Null Hypothesis Is Rejected")
```

## Ans of Sub-Q2:

No there is no significant difference between mean maths scores of two schools

# Task 5 Soln

```python
s1 = 12.3
n1 = 6
mean1 = 102.1

s2 = 7.52
n2 = 9
mean2 = 93.6

# Standard error of difference
SE = np.sqrt(s1**2/n1 + s2**2/n2)

# Welch degrees of freedom
df = (s1**2/n1 + s2**2/n2)**2 / (
    (s1**2/n1)**2/(n1-1) +
    (s2**2/n2)**2/(n2-1)
)

# 90% confidence -> 0.05 in each tail
t_critical = stats.t.ppf(0.95, df)

# Difference between means
difference = mean1 - mean2

# Margin of error
ME = t_critical * SE

lower = difference - ME
upper = difference + ME

print(lower, upper)
```
