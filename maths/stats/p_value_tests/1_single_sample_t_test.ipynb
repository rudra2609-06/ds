```python
import pandas as pd
import numpy as np
import scipy.stats as stats
```

```python
df = pd.read_csv('Titanic-Dataset.csv')
```

### Given Statement

```python
# H0 -> The mean age is 35
# H1 -> The mean age is less than 35
# Given , population mean (mu) -> 35
```

### Test Starts

```python
sample_age = df['Age'].dropna().sample(25).values
alpha = 0.05
```

```python
# As sample is only 25 and we need atleast 30 to prove that it follows normal distribution therefore we will first perform shapiro wilk test to check if the given sample follows the Normal distribution or not
```

```python
shapiro_age = stats.shapiro(sample_age)
shapiro_age
```

```python
# here p-value is greater than 0.05 (alpha) therefore it follows Normal distribution
```

```python
t_statistic,p_value = stats.ttest_1samp(sample_age,35)
```

```python
# as one sided
if p_value / 2 < alpha:
	print('Null Hypothesis Rejected')
else:
	print("Can't Null Hypothesis Rejected")
```

### Test Ends

### Proof

```python
# proof 
df['Age'].dropna().mean()

# it is less than 35
```

***When the sample size is less than 30 and it is not given that data follows normal distribution we can verify that using shapiro wilk test by comparing p_value with significance level.If it comes less than it does not follows Normal Distribution and if it is greater than it does follows Noraml Distribution.We just need to pass a column value on which we want to check***
