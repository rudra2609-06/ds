```python
df = pd.read_csv('Titanic-Dataset.csv')
```

```python
df.head()
```

```python
sample = df['Fare'].dropna().sample(30).to_numpy()
```

```python
sample_mean = np.mean(sample)
```

```python
sample_sigma = np.std(sample)
```

```python
SE = sample_sigma / (30 ** 0.5)
```

```python
t = 1 - 0.95
remaining_halves = t/2
t_value = 0.95 + remaining_halves
t_critical = stats.t.ppf(t_value,df = 29)
```

```python
lower_bound = sample_mean - (t_critical * SE)
upper_bound = sample_mean + (t_critical * SE)
```

```python
print(f'{lower_bound} - {upper_bound}')
```
