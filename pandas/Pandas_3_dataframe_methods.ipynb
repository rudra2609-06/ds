#### Handling NaN Values with `sort_values()`

When sorting, `NaN` (Not a Number) values by default appear at the bottom. The `na_position` parameter can be used to place them at the 'first' or 'last' position.

```python
# sort_values() with nan values

students = pd.DataFrame(
    {
        'name':['nitish','ankit','rupesh',np.nan,'mrityunjay',np.nan,'rishabh',np.nan,'aditya',np.nan],
        'college':['bit','iit','vit',np.nan,np.nan,'vlsi','ssit',np.nan,np.nan,'git'],
        'branch':['eee','it','cse',np.nan,'me','ce','civ','cse','bio',np.nan],
        'cgpa':[6.66,8.25,6.41,np.nan,5.6,9.0,7.4,10,7.4,np.nan],
        'package':[4,5,6,np.nan,6,7,8,9,np.nan,np.nan]

    }
)
students.sort_values('name') # if nan values available they will appear at bottom
students.sort_values('name',na_position='first') # if you want nan values at top by default bottom
```

### `rank()`

The `rank()` method is available for Series and assigns ranks to values in a Series. It can be used to determine the order of values, with options to handle ties and ascending/descending order.

```python
# rank (only in series)

runs = pd.read_csv('/content/batsman_runs_ipl.csv')
runs
```

#### Batsman Overall Rank

This section calculates the overall ranking of batsmen based on the total runs they have scored, with higher runs receiving a lower (better) rank.

```python
# batsman overall rank according to how many runs scored

ranking_batsman = runs['batsman_run'].rank(ascending=False) #by default ascending less runs -> more rank
ranking_batsman.head(10)
runs['batting_rank'] = ranking_batsman
runs.sort_values('batting_rank')
```

### `dropna()`

The `dropna()` method is used to remove missing values (NaN) from a Series or DataFrame. It offers parameters like `how` and `subset` to control the removal behavior.

```python
# dropna -> (series and dataframe) -> how = 'any' default

students['name'].dropna()

students.dropna() # if even a single row has missing row whole row delete
# i want to remove only those rows which has all values missing
students.dropna(how='all')

# remove those rows where name has nan
students.dropna(subset=['name'])

# remove those rows where name has nan or college
students.dropna(subset=['name','college']) # -> this is or operation
```

### `duplicated()` and `drop_duplicates()`

```python
# duplicated and drop_duplicates(keep='first(default)/last',subset='specific_col_name') -> series and Dataframe both

# duplicated helps in detecting duplicates row in dataframe or duplicate values in col/series whereas drop_duplicates() helps in removing those duplicates

student_data = {
    'name' : ['Rudra','Ansh','Heet','Rudra','Ram'],
    'iq' : [100,120,90,100,200],
    'lpa' : [10,20,9,10,30]
}
performance = pd.DataFrame(student_data)
performance.duplicated().sum() # one row is duplicated therefore gave True rest all are unique so it gave False

# performance.drop_duplicates(keep='last') # it keeps either first or last occurence by default drops last
# performance

```
