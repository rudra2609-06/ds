### Manual Filtering within Loops

### Highest Rated Movie of Each Genre

```python
df = pd.DataFrame()

for group, data in genres:
    temp = data[data['IMDB_Rating'] == data['IMDB_Rating'].max()]
    df = pd.concat([df, temp])

df
```

## The Apply Method (Split-Apply-Combine)

```python
# split -> apply -> combine stratergy
genres.apply(min)
```

### Custom Filtering with Apply

```python
# find number of movies starting with A for each group
def moviesWithA(group):
  return group[group['Series_Title'].str.startswith('A')]

genres.apply(moviesWithA)
```

### Custom Ranking with Apply

```python
# find ranking of each movie in the group according to IMDB score
def rankingAccorToImdb(group):
  group['genre_rank'] = group['IMDB_Rating'].rank(method='first',ascending=False)
  return group

genres.apply(rankingAccorToImdb)
```

### Group-wise Normalization

```python
# find normalized IMDB rating group wise
def normal(group):
  rating = group['IMDB_Rating']
  group['norm_rating'] = ((rating - rating.min()) / (rating.max() - rating.min()))
  return group

genres.apply(normal)
```
