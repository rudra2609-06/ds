Editing Series

```python
marks = {
	'hindi':100,
	'sst' : 90,
	'maths':20
}

my_marks = pd.Series(marks)
```

```python
# using indexing
my_marks['hindi'] = 40
my_marks
my_marks['gujarati'] = 60 #if that idx value not there it adds new
my_marks
```

```python
# slicing
my_marks
# change marks of sst and maths 1,2 idx
my_marks[1:3] = [91,21]
my_marks
```

```python
# fancy indexing
my_marks[[0,1]] = [0,0]
my_marks
```
