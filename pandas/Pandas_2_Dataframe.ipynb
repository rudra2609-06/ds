**Filtering In DataFrame**

```python
# find all the final winners -> ipl dataset
ipl[ipl['MatchNumber'] == 'Final'][['Season','WinningTeam']]
# ipl['WinningTeam']
```

```python
# how many superover finishes has occured
(ipl['SuperOver'] == 'Y').sum()
# we cannot use size after masking because it generally gives cols count we want rows
```

```python
# how many matches does csk won in kolkata
ipl[(ipl['WinningTeam'] == 'Chennai Super Kings') & (ipl['City'] == 'Kolkata')].shape[0] #shape returns -> (rows,cols)
```

```python
# toss winner is match winner or not in percentage
ipl
# mean is choosen here because it gives average true values
(ipl['TossWinner'] == ipl['WinningTeam']).mean() * 100
```

```python
# movies with ratings higher than 8 and votes > 10000
movies[(movies['imdb_rating'] > 8) & (movies['imdb_votes'] > 10000)]
```

```python
# Action movies rating higher than 7.5
# movies.columns

# movies[(movies['genres'].str.split('|').apply(lambda x : 'Action' in x)) & (movies['imdb_rating'] > 7.5)]
movies[(movies['genres'].str.contains('Action')) & (movies['imdb_rating'] > 7.5)]

```
