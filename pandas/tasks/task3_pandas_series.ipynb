###`Q-6:` Make a dataframe of each team in IPL with details like - Team Name, Matches Played, Win%, Home Win%, Away Win%.

Replace old team name as new name before performing any tasks.
```
Delhi Daredevils ->Delhi Capitals

Rising Pune Supergiants -> Rising Pune Supergiant
```

Note: Team1 represents Home team. Exclude not result matches.


````python
# code here
ipl_matches.head()

# ----------- Renaming Teams -----------

old_names = ['Delhi Daredevils','Kings XI Punjab','Rising Pune Supergiants']
new_names = ['Delhi Capitals','Punjab Kings','Rising Pune Supergiant']

mapped_names = dict(zip(old_names,new_names))
mapped_names
ipl_matches.replace(mapped_names,inplace=True)

all_teams = ipl_matches['Team1'].unique()
all_teams

res = []

for each_team in all_teams:

  # ---------- Team Name ---------
  team_name = each_team
  # ---------- Matches Played ----
  matches_played = ((ipl_matches['Team1'] == each_team) | (ipl_matches['Team2'] == each_team)).sum()
  # ---------- Win % ----
  win_perc = (((ipl_matches['WinningTeam'] == each_team).sum()) / matches_played) * 100
  res.append([team_name,matches_played,win_perc])


team_df = pd.DataFrame(res,columns=['team_name','matches_played','win_perc'])
team_df







````

###`Q-9:` IPL Point Table

Make a function `point_table` which take `season` as parameter and show points table in non-ascendng order of points and in ascending order of team name.

For winning - 2 Ponits;
For loosing - 0 Point
For not result both team gets 1 points.

Make dataframe which will have
`TeamName` `MatchesPlayed` `MatchesWon` `NoResult` `Points`
make `TeamName` as index.

```
season parametr should be one of these->
['2022', '2021', '2020/21', '2019', '2018', '2017', '2016', '2015', '2014', '2013', '2012', '2011', '2009/10', '2009', '2007/08']
```


Output of two Top 2 in season 2022
```
TeamName    MatchesPlayed	MatchesWon	NoResult	Points

Gujarat Titans	    16	   12	       0	     24
Rajasthan Royals	  17	   10	       0	     20

```

````python
# code here
ipl_matches['Season']

def point_table(season):
  try:
    season_data = ipl_matches[ipl_matches['Season'] == season]
    teams_in_season = season_data['Team1'].unique()
    points_data = []

    for each_team in teams_in_season:
      TeamName = each_team
      MatchesPlayed = ((season_data['Team1'] == each_team) | (season_data['Team2'] == each_team)).sum()
      MatchesWon = (season_data['WinningTeam'] == each_team).sum()
      team_data = season_data[(season_data['Team1'] == each_team) | (season_data['Team2'] == each_team)]
      NoResult = (team_data['WinningTeam'].isna()).sum()
      Points = (MatchesWon * 2) + NoResult
      points_data.append([TeamName,MatchesPlayed,MatchesWon,NoResult,Points])

    points_data_df = pd.DataFrame(points_data,columns=['TeamName','MatchesPlayed','MatchesWon','NoResult','Points'])
    points_data_df = points_data_df.sort_values(['Points','TeamName'],ascending=[False,True])
    points_data_df.set_index('TeamName',inplace=True)
    return points_data_df
  except:
    raise KeyError('No Such Season Found')

res = point_table('2022')
print(res.head(2))

````

###`Q-10:` IPL Point Table cont.
Extend the above IPL Point Table with an extra column as `SeasonPosition`

Team below top 4 after sorting on `Points` and then on `TeamName` Will have same `SeasonPosition` as there rank. use rank function.

Teams in Top four will have `SeasonPosition` as:
```
    'Winner' - Team won final
    'Runner' - Team lost Final
    3 - Losing Team in Qualifier2
    4 - Losing Team in Eliminator
```

For changing value of pariticular cell use `df.at[row_index, col_label] = value`

Output of two Top 2 in season 2022. Your result should have all teams.
```
TeamName    MatchesPlayed	MatchesWon	NoResult	Points   SeasonPosition

Gujarat Titans	    16	   12	       0	     24         Winner
Rajasthan Royals	  17	   10	       0	     20         Runner

```

Note: If you try to chnage value of view of any dataframe a warnig will be shown. To avoid it, make a copy of the dataframe you want to change in by `df.copy()`

````python
# code here
res = point_table('2022')
res1 = res.copy()

top_four_teams = res1.head(4)
res1['SeasonPosition'] = top_four_teams['Points'].rank(ascending=False,method='first')

res1.loc[res1['SeasonPosition'] == 1, 'SeasonPosition'] = 'Winner'
res1.loc[res1['SeasonPosition'] == 2, 'SeasonPosition'] = 'Runner'
res1.loc[(res1['SeasonPosition'] == 3), 'SeasonPosition'] = 'Losing Team in Qualifier2'
res1.loc[(res1['SeasonPosition'] == 4), 'SeasonPosition'] = 'Losing Team in Eliminator'
res1


````
