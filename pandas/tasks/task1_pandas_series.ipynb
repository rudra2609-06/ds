###`Q-5.`Write a function to change the data type of given a column or a Series. Function takes series and data type as input, returns the converted series.
```
series = pd.Series([1,2,'Python', 2.0, True, 100])
change to float type data
```
Note: Read about pd.to_numeric()

````python
# code here
def change_elements_data_type(series, dataType):
    valid_types = ['integer', 'signed', 'unsigned', 'float']

    if dataType not in valid_types:
        raise ValueError(f"Invalid dataType '{dataType}'. Choose from {valid_types}")

    return pd.to_numeric(series, downcast=dataType, errors='coerce')

s1 = pd.Series([1,2,'Python',2.0,True,100])
res = change_elements_data_type(s1,'float')
res
````

###`Q-9`

    ii. Item price is given in $, so convert it to rupees without currency symbol.

    iv. Fill nan with mean of the series


How csv file looks

```
item_name	item_price
Chips and Fresh Tomato Salsa	$2.39
Izze	$3.39
Nantucket Nectar	$3.39
Chips and Tomatillo-Green Chili Salsa	$2.39
Chicken Bowl	$16.98

```

````python
# code here
items = pd.read_csv('/content/items.csv',index_col='item_name').squeeze()

# show number of nan values
items[pd.isna(items)].sum()

# converting price to rupee with changing symbol
# non_nan_prices = items[~(pd.isna(items))]
rupee_prices = (
    items
    .str.replace('$','',regex=False)
    .astype(float)
    .where(lambda x: x.isna(),lambda x : x.mul(90))
)
rupee_prices

# Make data type of newly made series as float.
# rupee_prices = rupee_prices.astype(float)


#  Fill nan with mean of the series
rupee_prices.fillna(rupee_prices.mean(),inplace=True)
````
