### `Q-4:` Perform `groupby` based on the `"Team"` column and then perform Z Normalization on top of the below columns of each group:
1. Passes
2. Passes Completed
3. Attempted Line Breaks
4. Completed Line Breaks

You have to make a python function named `z_normalization` which takes two arguments:

1. *group:* Every group that you have created
2. *cols_to_perform:* This parameter takes a list of columns on which you have to perform the Z-Normalization.

For this problem, you have to use th `apply()` method.

$\Large Z \ - \ Normalization = \frac{X_i - \mu}{std} $

After that find out the below values for each group:
- minimum "Passess"
- maximum "Passess"
- minimum "Yellow Cards"
- maximum "Yellow cards"
- average "Yellow Cards"
- maximum "Attempted Line Breaks"
- minimum "Attempted Line Breaks"
- standardard deviation of "Attempted Line Breaks"
- average Possession

````python
# code here
fifa_df = pd.read_csv("https://docs.google.com/spreadsheets/d/e/2PACX-1vT3D_x_4DS6d51LKJ7ze1sxT5WpV5uiSVOFYHLwBiGru6vFyVv5h5-83AwFjxWYiWfCDjDAaarHAV-k/pub?gid=0&single=true&output=csv")
fifa_df


team_groupby = fifa_df.groupby('Team')

fifa_df.columns

def z_normalization(group, cols):
    group = group.copy()
    # now normalize
    for each_col in cols:
        mean = group[each_col].mean()
        std = group[each_col].std()
        group[each_col] = (group[each_col] - mean) / std

    return group


cols = ['Passes', 'Passes Completed', 'Attempted Line Breaks', 'Completed Line Breaks']
z_normalized_df = team_groupby.apply(z_normalization, cols)
z_normalized_df

team_groupby.agg({
    'Passes' : ['min','max'],
    'Yellow Cards' : ['min','max','mean'],
    'Attempted Line Breaks' : ['min','max','std'],
    'Possession (%)' : 'mean'
})

````

###`Q-5:` Find batsman in below category-
* Highest score while chasing
* Best Strike rate while chasing and have faced 100+ balls


> Chasing mean team batting in second inning

````python
# --------------- Highest score while chasing -----------------

chasing_innings = ipl[ipl['innings'] == 2]
highest_score_while_chasing = chasing_innings.groupby(['batter','ID'])['batsman_run'].sum().sort_values(ascending=False).head(1)

# --------------- Method - 1 for second problem -------------------

balls_faced = chasing_innings.groupby('batter').size()
batters_faced_more_than_100_balls = balls_faced[balls_faced > 100].sort_values(ascending=False)

runs = chasing_innings.groupby('batter')['total_run'].sum().sort_values(ascending=False)

strike_rate = round((runs / balls_faced) * 100,2)
strike_rate = strike_rate[balls_faced > 100].sort_values(ascending=False).head(1)
strike_rate

# --------------- Method - 2 for second problem ---------------

temp_df = chasing_innings[~(chasing_innings.extra_type == "wides")]

temp_df = temp_df.groupby('batter').agg({
    'batsman_run' : 'sum',
    'ballnumber' : 'count'
})

temp_df['Strike Rate'] = (temp_df['batsman_run'] / temp_df['ballnumber']) * 100
temp_df[temp_df['ballnumber'] >= 100].sort_values(by='Strike Rate',ascending=False).head(1)
````

###`Q-6` Most Successful bowler against any batsman. Find that pair of bowler and batsman.
> Most Successful in terms of dissmissal. A bowler who have dissmissed any batsman most no of times. If any two pairs have same no of dissmisal, consider runs conceded by bowler to that batsman. Those who have concede lesser runs is more successful.

````python
# code here
stats = ipl.groupby(['batter', 'bowler']).agg(
    dismissals=('player_out', 'count'),
    runs_conceded=('total_run', 'sum')
)
stats = stats.sort_values(
	by=['dismissals','runs_conceded'],
	ascending=[False,True]
)
stats['Rank'] = range(1,len(stats) + 1)
stats
````

###`Q-8:` Make a dataframe for all batting pairs played together.
```
Batsman1 Batsman2 Runs Avg StrikeRate
```

> Just to ease this question you can count wide-balls for strike rate.

````python
# code here

pairs = ipl.groupby(['batter','non-striker']).agg({
		 "total_run" : "sum",
		 "ballnumber" : "count",
		 "isWicketDelivery" : "sum"
}).reset_index()

pairs["Batsman1"]  = pairs["batter"]
pairs["Batsman2"]  = pairs["non-striker"]
pairs["Runs"] = pairs['total_run']
pairs["StrikeRate"] = (pairs["Runs"] / pairs["ballnumber"]) * 100
pairs["Avg"] = pairs["Runs"] / pairs["isWicketDelivery"]
pairs.sort_values(by='StrikeRate',ascending=False)
pairs[["Batsman1","Batsman2","Runs","Avg","StrikeRate"]]




````
