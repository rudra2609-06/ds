### Figure level function -> `catplot`

```python
# stip plot -> one categorical and one numerical

sns.catplot(kind = 'strip', data = tips, x = 'day', y = 'total_bill')
sns.catplot(kind = 'strip', data = tips, x = 'day', y = 'total_bill',jitter=False)
# sns.stripplot(data = tips, x = 'day', y = 'total_bill')
```

```python
# jitter

sns.catplot(kind = 'strip', data = tips, x = 'day', y = 'total_bill',jitter=0.1)
sns.catplot(kind = 'strip', data = tips, x = 'day', y = 'total_bill',jitter=1)
```

```python
# swarmplot -> similar to striplot but shows how data is distributed

sns.catplot(data = tips, kind = 'swarm', x = 'day', y = 'total_bill',hue='sex')
```

### Boxplot

A boxplot is a standardized way of displaying the distribution of data based on a five number summary (“minimum”, first quartile [Q1], median, third quartile [Q3] and “maximum”). It can tell you about your outliers and what their values are. Boxplots can also tell you if your data is symmetrical, how tightly your data is grouped and if and how your data is skewed.

```python
# Box plot

# sns.catplot(kind = 'box', data = tips, x = 'sex', y = 'total_bill')
sns.catplot(kind = 'box', data = tips, x = 'day', y = 'total_bill', hue = 'sex')
```

```python
# on single column

sns.catplot(kind = 'box', data = tips, y = 'total_bill')
```

```python
# violin plot -> Box plot + kde plot

sns.catplot(data = tips, kind = 'violin', y = 'total_bill', x = 'day', hue = 'sex')
```

```python
# categorical esimate plots
```

```python
# barplot

# on an avg males total bill and on an avg females total bill
import numpy as np
# default mean
sns.catplot(data = tips, kind = 'bar', x = 'sex', y = 'total_bill',estimator=np.std)
```

When there are multiple observations in each category, it also uses bootstrapping to compute a confidence interval around the estimate, which is plotted using error bars

```python
# pointplot -> barplot - bars and connect the errorbars used to study difference b/w categories

sns.catplot(data = tips, kind = 'point', x = 'sex', y = 'total_bill',estimator=np.std,hue='smoker')
```

When there are multiple observations in each category, it also uses bootstrapping to compute a confidence interval around the estimate, which is plotted using error bars

```python
# countplot

sns.countplot(data = tips, x = 'sex', hue = 'day')
```

```python
# faceting using catplot
# default of catplot is stripplot
sns.catplot(data=tips,x = 'sex',y = 'total_bill', col='smoker',kind = 'box')
```

### Regression Plots

- regplot
- lmplot

In the simplest invocation, both functions draw a scatterplot of two variables, x and y, and then fit the regression model y ~ x and plot the resulting regression line and a 95% confidence interval for that regression.

```python
tips.head()

# axis level funtion
# hue parameter is not available
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip')
sns.regplot(data = tips, x = 'total_bill', y = 'tip')
```

```python
# lmplot -> figure level fn
sns.lmplot(data = tips,x = 'total_bill', y = 'tip',hue = 'sex')
```

```python
# residplot -> diff b/w actual and predicted value is residual

sns.residplot(data = tips,x = 'total_bill', y = 'tip')
```

```python
# Multiplots
```

```python
# facetgrid

g = sns.FacetGrid(data = tips,col='day',row='time')
# x -> sex and y -> total_bill
g.map(sns.scatterplot,'sex','total_bill')
g.add_legend()
```

```python
# Pair plot -> plot pair wise relationship
# first it automatically detect numerical cols
# then plots scatter plot b/w every possible col
# if plot against same then it will plot histogram
# diagonal se upar ya niche same he hai


iris

sns.pairplot(iris,hue='species')
```

```python
# Pair grid -> pair plot is special case of PairGrid

# g = sns.PairGrid(iris)
# g.map(sns.scatterplot)

# as diagonal items are plotted against its own therefore wee can customize which plot to choose
g = sns.PairGrid(iris)
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
```

```python
# map_diag -> map_upper -> map_lower

g = sns.PairGrid(iris)
g.map_diag(sns.histplot)
g.map_upper(sns.scatterplot)
g.map_lower(sns.kdeplot)
```

```python
# vars if you want to restricts cols

g = sns.PairGrid(iris,vars=['sepal_width','petal_width'])
g.map_diag(sns.histplot)
g.map_upper(sns.scatterplot)
g.map_lower(sns.kdeplot)
```

### JointGrid Vs Jointplot

```python
sns.jointplot(data = tips,x = 'total_bill',  y = 'tip',kind ='kde')
sns.jointplot(data = tips,x = 'total_bill',  y = 'tip',kind ='hist')
```

```python
# JointGrid

g = sns.JointGrid(data=tips,x='total_bill',y='tip')
g.plot(sns.scatterplot,sns.histplot)
```
