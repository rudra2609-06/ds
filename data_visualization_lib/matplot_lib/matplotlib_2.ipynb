```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# colored scatterplots

iris = pd.read_csv('/content/iris.csv')


# plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'])
# plt.xlabel('Sepal Length')
# plt.ylabel('Petal Length')

# if we want to show which dot represents which species like setosa, versicolor,etc you
# need to give number to them

iris['Species'] = iris['Species'].replace({'Iris-setosa' : 0,'Iris-versicolor' : 1,'Iris-virginica' : 2})

plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],alpha=0.9)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
```

```python
# Plot size

plt.figure(figsize=(15,7))

plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],alpha=0.9)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
```

```python
# Annotations

batter = pd.read_csv('/content/batter.csv')

batter.head()

sample_df = batter.head(100).sample(25,random_state = 5)
sample_df
```

```python

# as there are 25 batsman
# width and height
plt.figure(figsize=(18,10))
for i in range(sample_df.shape[0]):
  plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])

plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])
```

```python
# Horizontal and vertical lines

plt.figure(figsize=(18,10))
for i in range(sample_df.shape[0]):
  plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])

plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])
plt.axhline(130,color='red')
plt.axvline(30,color='red')
```

```python
# a diff way to plot graph in matplotlib

fig, ax = plt.subplots(figsize=(15,8)) # -> another way to plot graph

ax.scatter(batter['avg'],batter['strike_rate'])
ax.set_title('Something')
ax.set_xlabel('avg')
ax.set_ylabel('strike rate')

fig.show()
```

```python
# if we want to plot 2 graphs first b/w avg and strike rate and other b/w avg and runs

# if there are more graphs let's say 4 so in that case we need to provide co-ordinates as [0,0] for first graph and then for 2nd [0,1]
# in that particular case we will get 2d arr in axis that's why

fig,ax = plt.subplots(nrows=2,ncols=1,sharex=True) # 2 * 1 -> 2 graphs

ax[0].scatter(batter['avg'],batter['strike_rate'])
ax[0].set_title('Avg V/S Strike Rate')
ax[1].scatter(batter['avg'],batter['runs'])
ax[1].set_title('Avg V/S Runs')

fig.show()
```

```python
# 3D Scatter Plots

fig = plt.figure()

ax = plt.subplot(projection='3d')

ax.scatter3D(batter['runs'],batter['avg'],batter['strike_rate'])

fig.show()
```

```python
# 3d line Plot

x = [0,1,2]
y = [10,20,40]
z = [20,14,80]

fig = plt.figure()
ax = plt.subplot(projection='3d')
ax.plot3D(x,y,z,color = 'red')

fig.show()
```

```python
# 3d surface Plots

x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

xx,yy = np.meshgrid(x,y)

z = xx ** 2 + yy ** 2
z.shape
```

```python
fig = plt.figure(figsize=(15,10))
ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
# contour plot represents 3d graph to 2d plot

fig = plt.figure(figsize=(15,10))
ax = plt.subplot()

p = ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(p)



```

```python
# contour plot represents 3d graph to 2d plot

fig = plt.figure(figsize=(15,10))
ax = plt.subplot()

p = ax.contourf(xx,yy,z,cmap='viridis')
fig.colorbar(p)



```

```python
# Heat Map

balls = pd.read_csv('/content/IPL_Ball_by_Ball_2008_2022.csv')
balls.head()

df = balls[(balls['ballnumber'].isin([1,2,3,4,5,6])) & (balls['batsman_run'] == 6)]

grid = df.pivot_table(index='overs',columns='ballnumber',values='batsman_run',aggfunc='count')

plt.imshow(grid)
plt.yticks(balls['overs'].unique(),list(range(1,21)))
plt.xticks(np.arange(0,6),list(range(1,7)))
plt.colorbar()
```
