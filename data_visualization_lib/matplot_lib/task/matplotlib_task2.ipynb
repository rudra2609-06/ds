### `Problem 9:` Create a surface plot and the 2 types of the contour plots of the below equation.

$$z = sin(\sqrt{x^2 + y^2})$$


````python
# code here

x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

xx,yy = np.meshgrid(x,y)

z = np.sin( ( ((xx) ** 2) + ((yy) ** 2) )** 0.5 )

fig = plt.figure(figsize=(12,6))

ax = fig.add_subplot(projection='3d')
ax.plot_surface(xx,yy,z,cmap='viridis')

# ---- 1st type contour plot ----
ax = fig.add_subplot()

ax.contour(xx,yy,z)

# ---- 2nd type contour plot ----
ax = fig.add_subplot()

ax.contourf(xx,yy,z)
````

### `Problem 10:` Create a surface plot and the 2 types of the contour plots of the below equation.

$$z = tan(\log_2({x^2 + y^2})$$

````python

x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

xx,yy = np.meshgrid(x,y)

z = np.tan(np.log2(((xx ** 2) + (yy ** 2))))

fig = plt.figure()
ax = fig.add_subplot(projection='3d')

ax.plot_surface(xx,yy,z)

# ------- 1st contour -------

ax = fig.add_subplot()

ax.contour(xx,yy,z)

# ------- 2nd contour -------

ax.contourf(xx,yy,z)
````
