# Plotting Multiple Line Graphs
This cell generates a line plot showing the run contributions of Virat Kohli and Rohit Sharma across different IPL seasons.

```python
import matplotlib.pyplot as plt

# Section: Plotting Multiple Line Graphs
# This cell generates a line plot showing the run contributions of
# Virat Kohli and Rohit Sharma across different IPL seasons.

plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

# Adding Titles and Labels
This cell enhances the plot by adding a title to the graph and labels to the x and y axes.

```python
# labels

plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
plt.title('Rohit Sharma V/S Virat Kholi')
plt.xlabel('Season Of IPL')
plt.ylabel('Runs Scored In IPL')
```

# Customizing Line Aesthetics
This section customizes the appearance of the lines in the plot, including color, linestyle, and linewidth.

```python
# colors, line (width and style)

plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dashed',linewidth=3)
plt.plot(batsman['index'],batsman['RG Sharma'],color = 'black',linestyle='dotted',linewidth=2)

plt.title('Rohit Sharma V/S Virat Kholi')
plt.xlabel('Season Of IPL')
plt.ylabel('Runs Scored In IPL')

```

# Adding Markers to Lines
This cell adds markers to the data points on the lines for better visualization.

```python
# markers

plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dashed',linewidth=3,marker='D',markersize=10)
plt.plot(batsman['index'],batsman['RG Sharma'],color = 'black',linestyle='dotted',linewidth=2,marker='+',markersize=5)

plt.title('Rohit Sharma V/S Virat Kholi')
plt.xlabel('Season Of IPL')
plt.ylabel('Runs Scored In IPL')
```

# Adding a Legend
This cell adds a legend to differentiate between the multiple lines on the plot.

```python
# legend

plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dashed',linewidth=3,marker='D',markersize=10,label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color = 'black',linestyle='dotted',linewidth=2,marker='+',markersize=5,label='Rohit')

plt.title('Rohit Sharma V/S Virat Kholi')
plt.xlabel('Season Of IPL')
plt.ylabel('Runs Scored In IPL')

plt.legend() # -> this fn is required to call for labels of lines
```

# Controlling Axis Limits
This cell demonstrates how to manually set the limits for the y-axis of a plot.

```python
# limiting axes

price = [10,20,30,20,10000]
years =[2022,2023,2024,2025,2026]

plt.plot(years,price)
plt.ylim(0,1000)
```

# Adding a Grid to the Plot
This cell adds a grid to the plot for easier reading of values.

```python
# grid


plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dashed',linewidth=3,marker='D',markersize=10,label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color = 'black',linestyle='dotted',linewidth=2,marker='+',markersize=5,label='Rohit')

plt.title('Rohit Sharma V/S Virat Kholi')
plt.xlabel('Season Of IPL')
plt.ylabel('Runs Scored In IPL')

plt.legend()


plt.grid()
```

# Displaying the Plot
This cell explicitly calls `plt.show()` to display the generated plot.

```python
# show

plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dashed',linewidth=3,marker='D',markersize=10,label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color = 'black',linestyle='dotted',linewidth=2,marker='+',markersize=5,label='Rohit')

plt.title('Rohit Sharma V/S Virat Kholi')
plt.xlabel('Season Of IPL')
plt.ylabel('Runs Scored In IPL')

plt.legend()


plt.grid()

plt.show() # -> this is the fn which displays the graph in other text editors
```

# Scatter Plot from DataFrame
This section loads batsman data from a CSV file and plots a scatter graph showing the relationship between average runs and strike rate.

```python
# on df

batter = pd.read_csv('/content/batter.csv')
df = batter.head(50)
plt.scatter(df['avg'],df['strike_rate'])
plt.ylabel('Strike Rate')
plt.xlabel('Average Runs')
plt.title('Strike Rate V/S Average Runs For Top 50 Batsman')
plt.grid()
```

# Customizing Scatter Plot Point Size
This cell demonstrates how to vary the size of points in a scatter plot based on a third numerical variable (e.g., 'size' of the dining party from the 'tips' dataset).

```python
# size

df = sns.load_dataset('tips')
df

plt.scatter(df['total_bill'],df['tip'],s=df['size']*20)
```

# Alternative Scatter Plot Method using `plt.plot()`
This cell shows an alternative way to create a scatter plot using `plt.plot()` with a marker argument, which can sometimes be faster for large datasets.

```python
# plotting without plt.scatter() using plt.plot()
# faster

plt.plot(df['total_bill'],df['tip'],'o')
```

# Handling Overlapping Labels
This cell shows how to use `plt.xticks(rotation=...)` to prevent label overlap when category names are long.

```python
# a problem


children = [10,20,30,40,50,60]
colors = ['Red Red Red Red Red Red','Yellow','Pink','Blue','Black','White']

plt.bar(colors,children,color='black')
# names get overlap
plt.xticks(rotation=70)
```

# Stacked Bar Chart
Visualizing multiple numerical categories for the same group by stacking bars on top of each other using the `bottom` parameter.

```python
# stacked Bar Chart  in DESC adn then add upper all every time
df

plt.bar(df['batsman'],df['2017'])
plt.bar(df['batsman'],df['2016'],bottom=df['2017'])
plt.bar(df['batsman'],df['2015'],bottom=df['2016'] + df['2017'])


```
