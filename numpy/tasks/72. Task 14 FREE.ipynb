###`Q-1:` Find the nearest element in the array to a given integer.
```
Ex:-
a=23 and array - [10 17 24 31 38 45 52 59].
Nearest element is 24
```
Hint: Read about this function `argmin()`

````python
# code here
arr = np.array([10,17,24,31,38,45,52,59])
a = 23

diff = abs(arr - a)

idx = np.argmin(diff)
print(f'Nearest Element:{arr[idx]}')


````

###`Q-4:` Your are given an array which is havig some nan value. You job is to fill those nan values with most common element in the array.
```
arr=np.array([[1,2,np.nan],[4,2,6],[np.nan,np.nan,5]])

```

````python
# code here
arr = np.array([[1,2,np.nan],[4,2,6],[np.nan,np.nan,5]])

non_nan_values = arr[~np.isnan(arr)]

# we temporarity converted arr to int because bincount requires int not float and nan values generally deals with float
most_common = np.bincount(non_nan_values.astype(int)).argmax()

arr[np.isnan(arr)] = most_common

arr



````

### `Q-9:` Write a program implement Boradcasting Rule to check if two array can be added or not.
Given tuples of shapes.
```
shape of a- (3,2,2)
shape of b- (2,2)

check_broadcast(a, b) -> return Boolean (True if can broadcasted, False other wise.)
```

````python
# code here


a_shape = (3,2,2)
b_shape = (2,2)

def check_if_broadcasted(a,b):
	a = list(a)
	b = list(b)

	while len(a) < len(b):
		a.insert(0,1)
	while len(b) < len(a):
		b.insert(0,1)
	
	for i in range(len(a)):
		if a[i] != b[i] and a[i] != 1 and b[i] != 1:
			return False
	return True

res = check_if_broadcasted(a_shape,b_shape)
res
````
