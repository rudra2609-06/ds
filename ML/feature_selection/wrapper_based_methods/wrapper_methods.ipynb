### 1. Setup & Imports

```python
import numpy as np
import pandas as pd
import scipy.stats as stats
from sklearn.datasets import load_iris, load_wine
from sklearn.linear_model import LogisticRegression
from mlxtend.feature_selection import ExhaustiveFeatureSelector as EFS, SequentialFeatureSelector as SFS
```

### 2. Exhaustive Feature Selection (EFS) on Iris Dataset

```python
X,y = load_iris(return_X_y=True,as_frame=True)

lr = LogisticRegression(max_iter=1000)

sel = EFS(estimator=lr,max_features=4,scoring='accuracy',cv=5)

model = sel.fit(X,y)
```

```python
print(model.best_feature_names_)
print(model.best_score_)
print(model.subsets_)
```

### 3. Sequential Backward Selection (SBS) on Wine Dataset

```python

X,y = load_wine(return_X_y=True,as_frame=True)

lr = LogisticRegression()

# sel = SFS(estimator=lr,forward=False,cv = 5,k_features=(1,13), scoring='accuracy',n_jobs=-1, floating=False)
sel = SFS(estimator=lr,forward=False,cv = 5,k_features="best", scoring='accuracy',n_jobs=-1, floating=False)
model = sel.fit(X,y)
```

```python
model.k_feature_names_
model.k_score_
```

### 4. Sequential Forward Selection (SFS) on Wine Dataset

```python

X,y = load_wine(return_X_y=True,as_frame=True)

lr = LogisticRegression()

# sel = SFS(estimator=lr,forward=False,cv = 5,k_features=(1,13), scoring='accuracy',n_jobs=-1, floating=False)
sel = SFS(estimator=lr,forward=True,cv = 5,k_features="best", scoring='accuracy',n_jobs=-1, floating=False)
model = sel.fit(X,y)
```

```python
model.k_feature_names_
# model.k_score_
```

### Implementing RFE

```python
from sklearn.feature_selection import RFECV
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier

X,y = load_iris(return_X_y=True,as_frame=True)

rfc = RandomForestClassifier()

rfe = RFECV(estimator=rfc,min_features_to_select=1,cv=5)

model = rfe.fit(X,y)

model.ranking_
```
