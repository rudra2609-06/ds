```python
import pandas as pd
import statsmodels.api as sm
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/erkansirin78/datasets/refs/heads/master/Advertising.csv',index_col=0)
```

```python
df.head()
```

```python
X = df[['TV','Radio','Newspaper']]
y = df['Sales']
```

```python
model = sm.OLS(y,X).fit()
```

```python

print(model.summary())
```
