```python
import numpy as np
import pandas as pd
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
np.random.seed(42)

n = 200

# Input features
X1 = np.random.uniform(1, 10, n)
X2 = np.random.uniform(1, 10, n)

# Noise
noise = np.random.normal(0, 10, n)

# Non-linear + skewed target
target = (
    np.exp(0.3 * X1)
    + X2**2
    + noise
)

df = pd.DataFrame({
    'X1': X1,
    'X2': X2,
    'target': target
})

print(df.head())
```

```python
lr = LinearRegression()
```

```python
x_test,x_train,y_test,y_train = train_test_split(df[['X1','X2']],df['target'], random_state=42, test_size=0.2)
```

```python
lr.fit(x_train,y_train)
```

```python
y_pred = lr.predict(x_test)
```

### Residual Plot
### Check for Non-Linearity

```python
import matplotlib.pyplot as plt

residuals = y_test - y_pred
plt.scatter(residuals,y_pred) # this shows random ness around 0 therefore non-linearity exists clearly
```
