```python
import numpy as np
import pandas as pd
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
import numpy as np
import pandas as pd

np.random.seed(42)

n = 200

# Generate input features
X1 = np.random.uniform(1, 10, n)
X2 = np.random.uniform(1, 10, n)

# Small random noise
noise = np.random.normal(0, 5, n)

# Linear relationship
target = 4 * X1 + 7 * X2 + 10 + noise

# Create DataFrame
df = pd.DataFrame({
    'X1': X1,
    'X2': X2,
    'target': target
})

print(df.head())
```

```python
x_test,x_train,y_test,y_train = train_test_split(df[['X1','X2']],df['target'], random_state=42, test_size=0.2)
lr = LinearRegression()
lr.fit(x_train,y_train)
y_pred = lr.predict(x_test)
```

```python
# histogram of residual

import matplotlib.pyplot as plt
residual = y_test - y_pred
plt.hist(residual)
plt.show()
```

```python
# running shapiro wilk test to the normality of residual assumption
from scipy import stats

shapiro_res = stats.shapiro(residual)
shapiro_res

if shapiro_res.pvalue < 0.05:
  print('Assumption Failed')
else:
  print('Assumption Passed')
```
