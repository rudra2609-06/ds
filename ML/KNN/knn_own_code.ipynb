```python
import numpy as np
from collections import Counter

class OwnKNN:
    def __init__(self, n_neighbours):
        self.n_neighbours = n_neighbours

    def euclidean_distance(self, point1, point2):
        return np.sqrt(np.sum((np.array(point1) - np.array(point2)) ** 2))

    def predict(self, X_train, y_train, X_test):
        predictions = []

        for each_point in X_test:                      # one test point at a time
            distances = []

            for i in range(len(X_train)):              # compare with every training row
                dist = self.euclidean_distance(each_point, X_train[i])
                distances.append((dist, y_train[i]))   # store (distance, label)

            distances.sort(key=lambda x: x[0])         # smallest distance first
            k_nearest = distances[:self.n_neighbours]  # take k closest
            k_nearest_labels = [label for _, label in k_nearest]

            most_common = Counter(k_nearest_labels).most_common(1)[0][0]
            predictions.append(most_common)

        return predictions
```
