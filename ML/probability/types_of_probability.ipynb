```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/Titanic-Dataset.csv')
```

## Joint Probability Distribution

```python
pd.crosstab(df['Survived'],df['Pclass'],normalize='all')
```

## Marginal Probability Distribution

**NOTE**:-**Here,when we do margins=True after that
at last col and last row we get marginal probability along with join probability in inner cell**

```python
pd.crosstab(df['Survived'],df['Pclass'],normalize='all',margins=True)
```

## Conditional Probability (Considering Pclass As Already Occured Event And Now We Are Calculating Survived|Pclass)

```python
pd.crosstab(df['Survived'],df['Pclass'],normalize='columns')
```

## Conditional Probability (Considering Survived As Already Occured Event And Now We Are Calculating Pclass|Survived)

```python
pd.crosstab(df['Survived'],df['Pclass'],normalize='index')
```
